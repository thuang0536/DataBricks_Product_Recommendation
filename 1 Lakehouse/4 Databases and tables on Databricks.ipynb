```sql
-- Read-only: available audit events mentioning this schema.
SELECT event_time, user_identity.email AS actor, service_name, action_name,
       request_params, response
       FROM system.access.audit
       WHERE workspace_id = '438592845960543'
         AND event_date >= date_sub(current_date(), 365)
           AND instr(lower(to_json(request_params)), 'new_default') > 0
           ORDER BY event_time ASC
           LIMIT 100;
```
